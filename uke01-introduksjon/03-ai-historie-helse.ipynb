Kjør cellene ovenfra og ned. Setup-cellen installerer bare ekstra pakker når notebooken åpnes i Colab.


```python
# AI-og-helse: Colab bootstrap v1
import importlib.util
import os
import subprocess
import sys
from pathlib import Path

IN_COLAB = "google.colab" in sys.modules
NOTEBOOK_PACKAGES = []
COLAB_DATA_MODE = 'synthetic-or-inline'


def _has_import(import_name: str) -> bool:
    return importlib.util.find_spec(import_name) is not None


def ensure_packages(packages=NOTEBOOK_PACKAGES):
    """Install only notebook-specific packages when running in Colab."""
    if not IN_COLAB:
        return

    missing = [package for package, import_name in packages if not _has_import(import_name)]
    if missing:
        print("Installerer Colab-pakker:", ", ".join(missing))
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", *missing])
    else:
        print("Alle notebook-spesifikke Colab-pakker er tilgjengelige.")


def get_secret(name: str, *aliases: str):
    """Read secrets from environment/.env locally or Colab Secrets in Colab."""
    for key in (name, *aliases):
        value = os.getenv(key)
        if value:
            os.environ[name] = value
            return value

    if IN_COLAB:
        try:
            from google.colab import userdata
        except Exception:
            userdata = None

        if userdata is not None:
            for key in (name, *aliases):
                try:
                    value = userdata.get(key)
                except Exception:
                    value = None
                if value:
                    os.environ[name] = value
                    return value

    return None


def mount_drive_if_needed():
    """Mount Google Drive explicitly in notebooks that need persistent artifacts."""
    if not IN_COLAB:
        return None
    from google.colab import drive

    drive.mount("/content/drive")
    return Path("/content/drive/MyDrive")


ensure_packages()
print("Miljø:", "Google Colab" if IN_COLAB else "lokalt")
print("Datamønster:", COLAB_DATA_MODE)
```

## Tidslinje: Fra 1950 til i dag

```python
# Opprett historisk tidslinje
historiske_hendelser = pd.DataFrame({
    'År': [1950, 1966, 1972, 1976, 1986, 1995, 2012, 2016, 2018, 2020, 2023],
    'Hendelse': [
        'Turing Test',
        'ELIZA (terapi-chatbot)',
        'MYCIN utvikles',
        'INTERNIST-I',
        'DXplain lanseres',
        'Første robot-assisterte kirurgi',
        'Deep Learning revolusjon',
        'AlphaGo slår Go-mester',
        'FDA godkjenner AI for diabetisk retinopati',
        'COVID-19: AI for vaksine og diagnose',
        'ChatGPT og generativ AI'
    ],
    'Område': [
        'Teoretisk grunnlag',
        'Psykiatri/Terapi',
        'Diagnostikk',
        'Intern medisin',
        'Klinisk beslutningsstøtte',
        'Kirurgi',
        'Maskinlæring',
        'Strategispill',
        'Oftalmologi',
        'Pandemihåndtering',
        'Generativ AI'
    ],
    'Betydning': [
        'Grunnlaget for AI-testing',
        'Første AI-terapeut',
        'Pionér i medisinsk ekspertsystem',
        'Diagnostisk ekspertsystem',
        'Kommersiell medisinsk AI',
        'AI-assistert kirurgi',
        'Gjennombrudd i neural nettverk',
        'AI overlegent mennesker i komplekse spill',
        'Første FDA-godkjente autonome AI-diagnose',
        'AI akselererer medisinsk forskning',
        'AI som medisinsk assistent'
    ]
})
```

```python
print(f"📊 Totalt {len(historiske_hendelser)} viktige milepæler i AI-historie")
historiske_hendelser.head(10)
```

```python
# Visualiser tidslinjen

plt.figure(figsize=(15, 10))

# Definer farger for forskjellige områder
område_farger = {
    'Teoretisk grunnlag': '#FF6B6B',
    'Psykiatri/Terapi': '#4ECDC4',
    'Diagnostikk': '#45B7D1',
    'Intern medisin': '#96CEB4',
    'Klinisk beslutningsstøtte': '#FECA57',
    'Kirurgi': '#FF9FF3',
    'Maskinlæring': '#54A0FF',
    'Strategispill': '#5F27CD',
    'Oftalmologi': '#00D2D3',
    'Pandemihåndtering': '#FF6348',
    'Generativ AI': '#2ED573'
}

# Lag scatterplot
for område in historiske_hendelser['Område'].unique():
    data = historiske_hendelser[historiske_hendelser['Område'] == område]
    plt.scatter(data['År'], [1]*len(data), 
               c=område_farger[område], s=200, alpha=0.7, 
               label=område, edgecolors='black', linewidth=1)

# Legg til hendelsesnavn
for i, row in historiske_hendelser.iterrows():
    plt.annotate(row['Hendelse'], 
                (row['År'], 1), 
                xytext=(0, 20), 
                textcoords='offset points',
                ha='center', va='bottom',
                fontsize=9, rotation=45,
                bbox=dict(boxstyle='round,pad=0.3', 
                         facecolor=område_farger[row['Område']], 
                         alpha=0.3))

plt.ylim(0.5, 1.5)
plt.xlabel('År', fontsize=14, fontweight='bold')
plt.title('AI i Helsevesenet: Historisk Tidslinje', fontsize=16, fontweight='bold', pad=20)
plt.grid(True, alpha=0.3)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10)
plt.yticks([])  # Fjern y-aksen siden den ikke er relevant

# Legg til tidsperioder
plt.axvspan(1950, 1980, alpha=0.1, color='red', label='Tidlig AI-periode')
plt.axvspan(1980, 2010, alpha=0.1, color='orange', label='Ekspertsystem-æra')
plt.axvspan(2010, 2025, alpha=0.1, color='green', label='Moderne AI-revolusjon')

plt.tight_layout()
plt.show()
```

## 🔍 Nærmere titt på viktige milepæler

### 1950-1980: Grunnlaget legges

**1950 - Turing Test**
- Alan Turing foreslår en test for kunstig intelligens
- Grunnlaget for å vurdere om maskiner kan "tenke"
- Fortsatt relevant for å evaluere AI-systemer i dag

**1966 - ELIZA**
- Første "terapeut-chatbot" utviklet av Joseph Weizenbaum
- Simulerte en Rogerian psykoterapeut
- Viste både potensial og begrensninger ved AI i mental helse

**1972 - MYCIN**
- Ekspertsystem for antibiotika-behandling
- Kunne diagnostisere og foreslå behandling for bakterielle infeksjoner
- Presterte like godt som eksperter på sitt område

### 1980-2010: Ekspertsystemer og gradvis implementering

**1976 - INTERNIST-I**
- Omfattende diagnostisk system for intern medisin
- Kunne håndtere flere tusen sykdommer
- Demonstrerte kompleksiteten i medisinsk diagnose

**1986 - DXplain**
- Første kommersielle medisinske ekspertsystem
- Fortsatt i bruk i dag (oppdatert versjon)
- Viser viktigheten av kontinuerlig utvikling

**1995 - Robot-assistert kirurgi**
- da Vinci-systemet introduseres
- Revolusjonerer presisjon i kirurgi
- Kombinerer AI med robotikk

### 2010-i dag: AI-revolusjonen

**2012 - Deep Learning-gjennombruddet**
- Convolutional Neural Networks (CNN) vinner ImageNet
- Starter den moderne AI-revolusjonen
- Åpner for avansert medisinsk bildeanalyse

**2016 - AlphaGo**
- Google DeepMind's AI slår verdensmester i Go
- Demonstrerer AI's evne til strategisk tenkning
- Inspirerer AI-forskning i kompleks medisinsk beslutning

**2018 - FDA-godkjenning**
- Første autonome AI-system godkjent av FDA
- IDx-DR for diabetisk retinopati-screening
- Milepæl for AI som medisinsk enhet

**2020 - COVID-19 pandemien**
- AI accelererer vaksine-utvikling
- Hurtig diagnose fra bildedannelse
- Demonstrerer AI's verdi i kriser

**2023 - Generativ AI**
- ChatGPT og lignende modeller
- Nye muligheter for medisinsk skriving og analyse
- Paradigmeskifte i hvordan vi samhandler med AI

```python
# Analyser utviklingen over tid
utviklingsanalyse = historiske_hendelser.copy()
utviklingsanalyse['Tiår'] = (utviklingsanalyse['År'] // 10) * 10

# Tell hendelser per tiår
hendelser_per_tiår = utviklingsanalyse.groupby('Tiår').size()

plt.figure(figsize=(12, 6))
plt.bar(hendelser_per_tiår.index, hendelser_per_tiår.values, 
        width=8, alpha=0.7, color='steelblue', edgecolor='black')

plt.xlabel('Tiår', fontsize=12, fontweight='bold')
plt.ylabel('Antall AI-milepæler', fontsize=12, fontweight='bold')
plt.title('AI-utvikling i helse: Accelererende innovasjon', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)

# Legg til trendlinje
x_vals = list(hendelser_per_tiår.index)
y_vals = list(hendelser_per_tiår.values)
z = np.polyfit(x_vals, y_vals, 1)
p = np.poly1d(z)
plt.plot(x_vals, p(x_vals), "r--", alpha=0.8, linewidth=2, label='Trend')

plt.legend()
plt.tight_layout()
plt.show()

print("🚀 Observasjon: AI-utviklingen i helse akselererer kraftig fra 2010-tallet!")
```

## 🎯 Lærdommer fra historien

### Suksessfaktorer
1. **Spesialisering først**: Tidlige AI-systemer som MYCIN fokuserte på smale domener
2. **Iterativ forbedring**: DXplain har utviklet seg kontinuerlig siden 1986
3. **Regulatorisk godkjenning**: FDA-prosessen sikrer sikkerhet og effekt
4. **Kriser som katalysator**: COVID-19 akselererte AI-adopsjonen

### Utfordringer og fiaskoer
1. **Over-optimisme**: Mange tidlige systemer lovet mer enn de kunne levere
2. **Brukeraksept**: Leger var skeptiske til AI-anbefalinger
3. **Integrasjon**: Vanskelig å integrere AI i eksisterende arbeidsflyt
4. **Ansvar og etikk**: Hvem er ansvarlig når AI tar feil?

```python
# Opprett data for fremtidsvisualisering
fremtid_df = pd.DataFrame({
    'Område': [
        'Medisinsk bildediagnostikk',
        'Medikamentutvikling', 
        'Pasientmonitorering',
        'Kirurgisk assistanse',
        'Personlig medisin',
        'Diagnostisk støtte',
        'Administrativ automatisering',
        'Telemedisin'
    ],
    'Nåværende_nivå': [7, 4, 5, 3, 4, 6, 5, 6],  # AI-modenhet i 2024 (1-10 skala)
    'Forventet_2030': [9, 8, 8, 7, 8, 9, 8, 9]   # Forventet nivå i 2030
})
```

```python
# Vis dataene
print("📊 Forventet AI-utvikling i helse:")
print("="*50)
for idx, row in fremtid_df.iterrows():
    forbedring = row['Forventet_2030'] - row['Nåværende_nivå']
    print(f"🏥 {row['Område']}")
    print(f"   2024: {row['Nåværende_nivå']}/10 → 2030: {row['Forventet_2030']}/10 (+{forbedring})")
    print()
```

```python
# Visualiser forventet utvikling
fig, ax = plt.subplots(figsize=(12, 8))
x = np.arange(len(fremtid_df['Område']))
width = 0.35

plt.bar(x - width/2, fremtid_df['Nåværende_nivå'], width, 
        label='2024', alpha=0.8, color='lightcoral')
plt.bar(x + width/2, fremtid_df['Forventet_2030'], width, 
        label='2030 (forventet)', alpha=0.8, color='lightgreen')

plt.xlabel('Helseområder', fontweight='bold')
plt.ylabel('AI-modenhet (1-10 skala)', fontweight='bold')
plt.title('Forventet AI-utvikling i helse 2024-2030', fontweight='bold', pad=20)
plt.xticks(x, fremtid_df['Område'], rotation=45, ha='right')
plt.legend()
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()

# Beregn gjennomsnittlig vekst
gjennomsnitt_2024 = fremtid_df['Nåværende_nivå'].mean()
gjennomsnitt_2030 = fremtid_df['Forventet_2030'].mean()
total_vekst = gjennomsnitt_2030 - gjennomsnitt_2024

print("📈 Nøkkelstatistikk:")
print(f"   Gjennomsnittlig AI-modenhet 2024: {gjennomsnitt_2024:.1f}/10")
print(f"   Forventet modenhet 2030: {gjennomsnitt_2030:.1f}/10")
print(f"   Total vekst: +{total_vekst:.1f} poeng")
print()
print("🌟 Fremtiden: AI vil transformere alle aspekter av helsevesenet!")
```
